Main implementation

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
class Perceptron:
    def __init__(self,eta,epochs):
        self.waights = np.random.rand(3) * 1e-4
        print(f"Initial weights: {self.waights}")
        self.eta = eta
        self.epochs = epochs
    
    def activation_function(self,input,waights):
        z = np.dot(input, waights)
        return np.where(z >= 0.0, 1, 0)

    def fit(self,X,y):
        self.X = X
        self.y = y
        
        x_with_bias = np.c_[self.X, -np.ones((len(X), 1))]
        print(f"X with bias: {x_with_bias}")
        
        for epoch in range(self.epochs):
            print("Epoch:", epoch)
        
            y_hat = self.activation_function(x_with_bias, self.waights)
            print(f"Predicted values: {y_hat}")
        
            error = self.y - y_hat
            print(f"Error: {error}")
        
            #waights update
            self.waights += self.eta * np.dot(x_with_bias.T, error)
            print(f"Updated weights: {self.waights}")
        
        
            print("#############################")
            
    def predict(self,X):
        X_with_bias = np.c_[X, -np.ones((len(X), 1))]
        
        return self.activation_function(X_with_bias, self.waights)
```

```python
import pandas as pd
data = pd.DataFrame({
    "x1": [1, 2, 3, 4, 5],
    "x2": [5, 4, 3, 2, 1],
    "y": [0, 1, 0, 1, 0]
})

AND = pd.DataFrame(data)

x = AND.drop("y", axis=1)
y = AND["y"]
y.to_frame()
```

#model Prediction

```python
model = Perceptron(eta=0.5, epochs=10)
model.fit(x,y)
```
